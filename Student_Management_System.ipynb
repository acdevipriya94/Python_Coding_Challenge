```python
students = {}

def add_stu():
  name = input("Enter Student Name : ")
  age = int(input("Enter Student Age : "))
  grade = int(input("Enter Student Grade : "))
  students[name] = {"Age" : age , "Grade" : grade}
```

```python
def del_stu():
  name = input("Enter student name to delete : ")
  if name in students:
    del students[name]
  else:
    print("Student not found")
```

```python
def update_stu():
  name = input("Enter Student name to update : ")
  if name in students:
    students[name]["Grade"] = int(input("Enter New Grade : "))
  else:
    print("Student not found")
```

```python
def search_stu():
  name = input("Enter Student name to search : ")
  if name in students:
    print(students[name])
  else:
    print("Student not found")
```

```python
def show_all_stu():
  for name in students:
    print(name, students[name])
```

```python
while True:
  print("/n 1.Add 2.Delete 3.Update 4.Search 5.Show_All 6.Exit")
  choice = int(input("Enter Your Choice"))

  if choice == 1:
    add_stu()
  elif choice == 2:
    del_stu()
  elif choice == 3:
    update_stu()
  elif choice == 4:
    search_stu()
  elif choice == 5:
    show_all_stu()
  elif choice == 6:
    break
  else:
    print("Invalid Choice")

```
