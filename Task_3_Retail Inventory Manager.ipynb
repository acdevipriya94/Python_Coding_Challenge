```python
categories = ("Electronics" , "Stationary" , "Clothing")
products = ["laptop" , "Notebook" , "Salwar"]
new_product = input("Enter a new product : ")
products.append(new_product)
products
```

```python
index = int(input("Enter the index of the product to remove from list : "))
products.pop(index)
print(f"Final Categories : {categories}")
print(f"Final Products : {products}")
```
