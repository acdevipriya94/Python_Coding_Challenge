```python
total = 0
count = 0
while True:
  price = float(input("Enter price ( 0 to stop ) :  "))
  if price == 0:
    break
  elif price < 0:
    print("Invalid Price Entered !!!")
    continue
  total = total + price
  count = count + 1

if total >= 1000:
  discount = total * 0.2
elif total >= 500:
  discount = total * 0.1
elif total >= 200:
  discount = total * 0.05
else:
  discount = 0

final_price = total - discount

print(f"Total Items : {count}")
print(f"Subtotal : {total}")
print(f"Discount Amount : {discount}")
print(f"Final Total : {final_price}")
```
