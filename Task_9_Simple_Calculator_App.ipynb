```python
def add(a,b):
  return a+b
def sub(a,b):
  return a-b
def mul(a,b):
  return a*b
def div(a,b):
  if b == 0:
    return "Cant divide by Zero"
  return a/b

a= float(input("Enter first number : "))
b= float(input("Enter second number : "))
ope = input("Enter the operation needs to be performned (+,-,*,/) : ")

if ope == "+":
  result = add(a,b)
elif ope == "-":
  result = sub(a,b)
elif ope == "*":
  result = mul(a,b)
elif ope == "/":
  result = div(a,b)
else:
  print("Invalid Operator !!!")

print(f"Result : {result}")
```

```python
if isinstance(result, (int,float)):
  print(f"Absolute : {abs(result)}")
  print(f"Rounded Result : {round(result , 2)}")
  print(f"Power of Result : {pow(result, 2)}")
else:
  print(result)
```
