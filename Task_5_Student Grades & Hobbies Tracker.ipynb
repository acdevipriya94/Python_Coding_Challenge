```python
grades = {}
for i in range(3):
  name = input("Enter Student Name : ")
  grade = int(input("Enter Student Grade : "))
  grades[name] = grade
print("Grades = ", grades)
```

```python
hobbies = set()
for i in range(3):
  hobby = input("Enter a Hobby : ")
  hobbies.add(hobby)

frozon_hobbies = frozenset(hobbies)

print(f"Hobbies : {hobbies}")
print(f"Frozen_Hobbies : {frozon_hobbies}")
```
