```python
import random
scores = [random.randint(1,100) for i in range (10)]

for index, score in enumerate(scores , start = 1):
  print(index, ":" , score)
```

```python
highest_score = []
for s in scores:
  if s > 75:
    highest_score.append(s)
```

```python
highest_score
```

```python
avg = sum(scores) /len(scores)
print(avg)
```

```python
print(f"Scores Greater than 75 : {highest_score}")
print(f"Average Score : {avg}")
```
